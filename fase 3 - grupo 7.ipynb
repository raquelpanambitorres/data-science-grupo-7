### 1. Preparación de Datos y Definición de la Variable Objetivo

Para predecir de forma realista el impacto socioeconómico sin el sesgo biológico de la edad, definiremos una variable objetivo llamada `rezago_escolar`.
Un estudiante presenta **rezago escolar (1)** si el nivel formal que aprobó (`nivel_grado_aprobado`) es inferior al que teóricamente le corresponde para su edad actual, o si ya no asiste actualmente a la educación formal habiendo abandonado los estudios antes de tiempo. Caso contrario, se considera **sin rezago (0)**.

````python
import numpy as np
import pandas as pd

# Aseguramos la existencia de 'pob', 'ingrefam' y 'viv' si el kernel se reinició
if 'pob' not in globals():
    print("Re-inicializando el DataFrame 'pob'...")
    pob = poblacion_datos[['UPM', 'NVIVI', 'NHOGA', 'P02', 'P03', 'ED01', 'ED02', 'ED03', 'ED0504', 'ED06C', 'ED08', 'ED09', 'ED10', 'S01A', 'S01B' ]].copy()
    pob.rename(columns={
        'UPM': 'upm', 'NVIVI': 'vivienda', 'NHOGA': 'hogar', 'P02': 'edad', 'P03': 'parentesco',
        'ED01': 'idioma_casa', 'ED02': 'sabe_leer_escribir', 'ED03': 'asiste_educ_formal',
        'ED0504': 'nivel_grado_aprobado', 'ED06C': 'titulo_diploma', 'ED08': 'asiste_educ_formal_actual',
        'ED09': 'sector_educ_formal', 'ED10': 'motivo_no_asiste', 'S01A': 'seguro_medico_pais', 'S01B': 'seguro_medico_local'
    }, inplace=True)
    for col in ['idioma_casa', 'sabe_leer_escribir', 'asiste_educ_formal', 'nivel_grado_aprobado', 'titulo_diploma', 'asiste_educ_formal_actual', 'sector_educ_formal', 'motivo_no_asiste', 'seguro_medico_local', 'edad']:
        pob[col] = pob[col].replace(r'^\s*$', float('nan'), regex=True)
        pob[col] = pd.to_numeric(pob[col], errors='coerce')

if 'ingrefam' not in globals():
    print("Re-inicializando el DataFrame 'ingrefam'...")
    ingrefam = ingrefam_datos[['upm', 'nvivi', 'nhoga', 'ingrem', 'pobrezai', 'pobnopoi']].copy()
    ingrefam.rename(columns={
        'upm': 'upm', 'nvivi': 'vivienda', 'nhoga': 'hogar', 'ingrem': 'ing_familiar', 'pobrezai': 'est_pobreza', 'pobnopoi': 'cond_pobreza'
    }, inplace=True)
    ingrefam['ing_familiar'] = ingrefam['ing_familiar'].str.replace(',', '.', regex=False).astype(float)

if 'viv' not in globals():
    print("Re-inicializando el DataFrame 'viv'...")
    viv = vivienda_datos[['UPM', 'NVIVI','NHOGA' ,'DPTO','AREA', 'HOMBRES', 'MUJERES', 'TOTAL', 'THOGAV', 'POBREZAI', 'POBNOPOI']].copy()
    viv.rename(columns={
        'UPM': 'upm', 'NVIVI': 'vivienda', 'NHOGA': 'hogar', 'DPTO': 'departamento', 'AREA': 'area',
        'HOMBRES': 'hombres', 'MUJERES': 'mujeres', 'TOTAL': 'total', 'THOGAV': 'tipo_hogar', 'POBREZAI': 'est_pobreza', 'POBNOPOI': 'cond_pobreza'
    }, inplace=True)

# 1. Realizamos la fusión completa de características del hogar, económicas e individuales
pob_completo = pd.merge(
    pob,
    ingrefam[['upm', 'vivienda', 'hogar', 'ing_familiar', 'est_pobreza', 'cond_pobreza']],
    on=['upm', 'vivienda', 'hogar'],
    how='inner'
)
pob_completo = pd.merge(
    pob_completo,
    viv[['upm', 'vivienda', 'hogar', 'departamento', 'area', 'total', 'tipo_hogar']],
    on=['upm', 'vivienda', 'hogar'],
    how='inner'
)

# 2. Filtramos para el rango de niños y jóvenes en edad escolar activa (por ejemplo, de 6 a 17 años)
jovenes_ml = pob_completo[(pob_completo['edad'] >= 6) & (pob_completo['edad'] <= 17)].copy()

# 3. Definición de Rezago Escolar como Target (Y)
def calcular_anios_estudio(row):
    nivel = row['nivel_grado_aprobado']
    if pd.isna(nivel) or nivel == 9999 or nivel == 8888 or nivel == 0:
        return 0
    if 301 <= nivel <= 306:
        return nivel - 300
    elif 407 <= nivel <= 409:
        return nivel - 400
    elif 501 <= nivel <= 1604:
        digito_final = nivel % 10
        return 9 + min(digito_final, 3)
    else:
        return 0

jovenes_ml['anios_estudio'] = jovenes_ml.apply(calcular_anios_estudio, axis=1)

# Un estudiante tiene rezago si: edad - 7 > anios_estudio
jovenes_ml['rezago_escolar'] = ((jovenes_ml['edad'] - 7) > jovenes_ml['anios_estudio']).astype(int)

# Mostrar la distribución del Target
print("Distribución de la variable objetivo (Rezago Escolar: 1 = Sí, 0 = No):")
print(jovenes_ml['rezago_escolar'].value_counts(normalize=True) * 100)

# Selección de features de interés
features = [
    'edad', 'idioma_casa', 'sabe_leer_escribir', 'asiste_educ_formal_actual',
    'ing_familiar', 'est_pobreza', 'cond_pobreza', 'departamento', 'area',
    'total', 'tipo_hogar', 'seguro_medico_pais'
]

# Filtrar el dataset final
dataset_ml = jovenes_ml[features + ['rezago_escolar']].dropna(subset=['ing_familiar', 'rezago_escolar']).copy()
display(dataset_ml.head())
````

### 2. Entrenamiento y Evaluación de Modelos de Machine Learning

A continuación, construiremos el Pipeline de preprocesamiento utilizando `ColumnTransformer`, dividiremos los datos de manera estratificada y entrenaremos clasificadores de Machine Learning para comparar su capacidad de predicción del rezago escolar.

````python
# 1. Separar variables independientes (X) y variable objetivo (y)
X = dataset_ml[features]
y = dataset_ml['rezago_escolar']

# 2. Identificar tipos de columnas para el preprocesamiento
numeric_features = ['edad', 'ing_familiar', 'total']
categorical_features = ['idioma_casa', 'sabe_leer_escribir', 'asiste_educ_formal_actual', 'est_pobreza', 'cond_pobreza', 'departamento', 'area', 'tipo_hogar', 'seguro_medico_pais']

# 3. Dividir en conjuntos de entrenamiento y prueba (estratificado debido al desbalance)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 4. Crear transformadores para el Pipeline
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

# 5. Definir los modelos a evaluar
modelos = {
    'Clasificador Base (Dummy)': DummyClassifier(strategy='stratified', random_state=42),
    'Regresión Logística': LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42),
    'Random Forest': RandomForestClassifier(class_weight='balanced', random_state=42, n_estimators=100, max_depth=8),
    'Gradient Boosting (Hist)': HistGradientBoostingClassifier(class_weight='balanced', random_state=42)
}

# 6. Entrenar y evaluar cada modelo
resultados = []

for nombre, modelo in modelos.items():
    pipeline_modelo = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', modelo)
    ])

    # Entrenar
    pipeline_modelo.fit(X_train, y_train)

    # Predicciones
    y_pred = pipeline_modelo.predict(X_test)
    y_proba = pipeline_modelo.predict_proba(X_test)[:, 1] if hasattr(pipeline_modelo, "predict_proba") else np.zeros_like(y_pred)

    # Calcular métricas
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_proba)

    resultados.append({
        'Modelo': nombre,
        'Accuracy': acc,
        'Precisión': prec,
        'Recall': rec,
        'F1-Score': f1,
        'ROC-AUC': auc
    })

# Presentar resultados consolidados
df_resultados = pd.DataFrame(resultados)
display(df_resultados.sort_values(by='ROC-AUC', ascending=False))
````

Esto nos permitirá contrastar las predicciones del modelo con las hipótesis planteadas.

````python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.inspection import permutation_importance
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import train_test_split

# 1. Asegurar la preparación de los conjuntos de datos de forma local para robustez
numeric_features = ['edad', 'ing_familiar', 'total']
categorical_features = ['idioma_casa', 'sabe_leer_escribir', 'asiste_educ_formal_actual', 'est_pobreza', 'cond_pobreza', 'departamento', 'area', 'tipo_hogar', 'seguro_medico_pais']

X = dataset_ml[features]
y = dataset_ml['rezago_escolar']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

local_preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

# 2. Definir y ajustar el pipeline del mejor modelo de forma local
best_pipeline = Pipeline(steps=[
    ('preprocessor', local_preprocessor),
    ('classifier', HistGradientBoostingClassifier(class_weight='balanced', random_state=42))
])
best_pipeline.fit(X_train, y_train)

# 3. Calcular la importancia de variables por permutación
result_importance = permutation_importance(
    best_pipeline, X_test, y_test, n_repeats=10, random_state=42, n_jobs=-1
)

# 4. Formatear y ordenar los resultados
importances_df = pd.DataFrame({
    'Feature': features,
    'Importance': result_importance.importances_mean
}).sort_values(by='Importance', ascending=True)

# 5. Graficar la importancia de las variables
plt.figure(figsize=(10, 8))
plt.barh(importances_df['Feature'], importances_df['Importance'], color='teal', edgecolor='black')
plt.title('Importancia de Variables por Permutación (Test Set) - Rezago Escolar', fontsize=14, fontweight='bold')
plt.xlabel('Reducción media en la métrica del modelo (Score)', fontsize=12)
plt.ylabel('Variables del Hogar / Individuales', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

Optimización mediante GridSearchCV
Para cumplir formalmente con la validación cruzada y el ajuste sistemático de hiperparámetros sin incurrir en fuga de datos (data leakage), creamos un proceso de búsqueda en grilla utilizando `GridSearchCV` integrado con el Pipeline del modelo campeón.

````python
from sklearn.model_selection import GridSearchCV

# 1. Definir el Pipeline base para el modelo campeón
pipeline_opt = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', HistGradientBoostingClassifier(class_weight='balanced', random_state=42))
])

# 2. Definir una grilla de hiperparámetros para evaluar
# Evaluamos combinaciones de tasa de aprendizaje y profundidad del árbol
param_grid = {
    'classifier__learning_rate': [0.01, 0.05, 0.1],
    'classifier__max_depth': [3, 5, 8],
    'classifier__max_iter': [50, 100]
}

# 3. Configurar GridSearchCV con validación cruzada de 5 folds (Stratified por defecto para clasificación)
grid_search = GridSearchCV(
    estimator=pipeline_opt,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=5,
    n_jobs=-1,
    verbose=1
)

# 4. Ajustar el buscador de grilla ÚNICAMENTE en el set de entrenamiento
print("Iniciando la búsqueda en grilla sistemática con validación cruzada...")
grid_search.fit(X_train, y_train)

# 5. Reportar los mejores parámetros encontrados
print("\n--- Mejores Hiperparámetros Encontrados ---")
print(grid_search.best_params_)
print(f"Mejor score ROC-AUC obtenido en Validación Cruzada: {grid_search.best_score_:.4f}")

# 6. Evaluar el modelo final optimizado en el conjunto de prueba (una sola vez)
y_pred_opt = grid_search.predict(X_test)
y_proba_opt = grid_search.predict_proba(X_test)[:, 1]

print("\n--- Desempeño del Modelo Optimizado en el Conjunto de Prueba final ---")
print(f"Accuracy: {accuracy_score(y_test, y_pred_opt):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred_opt):.4f}")
print(f"ROC-AUC final: {roc_auc_score(y_test, y_proba_opt):.4f}")
````

*   **Sesgo Algorítmico y Discriminación Sistémica:** El modelo utiliza variables altamente sensibles como `idioma_casa` (idioma mayoritario hablado en el hogar: Guaraní vs. Castellano) o el departamento geográfico. *   **La trampa de la causalidad:** El modelo predice correlaciones estadísticas potentes en el espacio de características del hogar, pero no explica las causas psicopedagógicas individuales. Tratar la pobreza puramente como un factor predictivo numérico deshumaniza las realidades particulares de los estudiantes.
