# Importación y limpieza de datos

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import numpy as np
````

````python
# INGREFAM_EPHC_ANUAL_2025.csv
# https://drive.google.com/file/d/1_OotStgeP-dmqMvsJt9qPNvI0Y9bgZoV/view?usp=drive_link
# VIVIENDA_REG01_EPHC_ANUAL_2025.csv
# https://drive.google.com/file/d/17bktcClbDb4as8mK779Ct_qFLioYqYs9/view?usp=sharing
# POBLACION_REG02_EPHC_ANUAL_2025.csv
# https://drive.google.com/file/d/1DcY_xHywuTkclnzx6mqhWHU2gdZtt9EU/view?usp=sharing

file_id_INGREFAM = '1_OotStgeP-dmqMvsJt9qPNvI0Y9bgZoV'
file_id_VIVIENDA = '17bktcClbDb4as8mK779Ct_qFLioYqYs9'
file_id_POBLACION = '1DcY_xHywuTkclnzx6mqhWHU2gdZtt9EU'

# Download direct link
direct_link_INGREFAM = f'https://drive.google.com/uc?export=download&id={file_id_INGREFAM}'
direct_link_VIVIENDA = f'https://drive.google.com/uc?export=download&id={file_id_VIVIENDA}'
direct_link_POBLACION = f'https://drive.google.com/uc?export=download&id={file_id_POBLACION}'

# Read direct link into pandas
ingrefam_datos = pd.read_csv(direct_link_INGREFAM, sep=';', on_bad_lines='skip')
vivienda_datos = pd.read_csv(direct_link_VIVIENDA, sep=';', on_bad_lines='skip')
poblacion_datos = pd.read_csv(direct_link_POBLACION, sep=';', on_bad_lines='skip', low_memory=False)
````

````python
ingrefam_datos.info()
````

````python
# Selección, renombrado y casteo correcto de ingrefam
ingrefam = ingrefam_datos[['upm', 'nvivi', 'nhoga', 'ingrem', 'pobrezai', 'pobnopoi']].copy()
dic_ingr = {
    'upm': 'upm',
    'nvivi': 'vivienda',
    'nhoga': 'hogar',
    'ingrem': 'ing_familiar',
    'pobrezai': 'est_pobreza',
    'pobnopoi': 'cond_pobreza'
}
ingrefam.rename(columns=dic_ingr, inplace=True)

# Convertir 'ing_familiar' a flotante usando punto como separador decimal
ingrefam['ing_familiar'] = ingrefam['ing_familiar'].str.replace(',', '.', regex=False).astype(float)

display(ingrefam.head())
````

````python
# Selección y renombrado correcto de vivienda_datos
viv = vivienda_datos[['UPM', 'NVIVI','NHOGA' ,'DPTO','AREA', 'HOMBRES', 'MUJERES', 'TOTAL', 'THOGAV', 'POBREZAI', 'POBNOPOI']].copy()
dic_viv = {
    'UPM': 'upm',
    'NVIVI': 'vivienda',
    'NHOGA': 'hogar',
    'DPTO': 'departamento',
    'AREA': 'area',
    'HOMBRES': 'hombres',
    'MUJERES': 'mujeres',
    'TOTAL': 'total',
    'THOGAV': 'tipo_hogar',
    'POBREZAI': 'est_pobreza',
    'POBNOPOI': 'cond_pobreza'
}
viv.rename(columns=dic_viv, inplace=True)

display(viv.head())
````

````python
# Selección, renombrado y conversión de tipos correcta de poblacion_datos
pob = poblacion_datos[['UPM', 'NVIVI', 'NHOGA', 'P02', 'P03', 'ED01', 'ED02', 'ED03', 'ED0504', 'ED06C', 'ED08', 'ED09', 'ED10', 'S01A', 'S01B' ]].copy()
dic_pob = {
    'UPM': 'upm',
    'NVIVI': 'vivienda',
    'NHOGA': 'hogar',
    'P02': 'edad',
    'P03': 'parentesco',
    'ED01': 'idioma_casa',
    'ED02': 'sabe_leer_escribir',
    'ED03': 'asiste_educ_formal',
    'ED0504': 'nivel_grado_aprobado',
    'ED06C': 'titulo_diploma',
    'ED08': 'asiste_educ_formal_actual',
    'ED09': 'sector_educ_formal',
    'ED10': 'motivo_no_asiste',
    'S01A': 'seguro_medico_pais',
    'S01B': 'seguro_medico_local'
}
pob.rename(columns=dic_pob, inplace=True)

# Convertir columnas que tienen tipos mixtos o strings vacíos a numérico
columns_to_convert_to_numeric = [
    'idioma_casa',
    'sabe_leer_escribir',
    'asiste_educ_formal',
    'nivel_grado_aprobado',
    'titulo_diploma',
    'asiste_educ_formal_actual',
    'sector_educ_formal',
    'motivo_no_asiste',
    'seguro_medico_local',
    'edad'
]

for col in columns_to_convert_to_numeric:
    pob[col] = pob[col].replace(r'^\s*$', float('nan'), regex=True)
    pob[col] = pd.to_numeric(pob[col], errors='coerce')

display(pob.head())
````

````python
# Fusionar `pob` con `ingrefam` para obtener los datos económicos del hogar.
pob_ingrefam = pd.merge(pob, ingrefam[['upm', 'vivienda', 'hogar', 'ing_familiar', 'est_pobreza', 'cond_pobreza']], on=['upm', 'vivienda', 'hogar'], how='left')

# Filtrar para incluir solo a los niños (menores de 18 años, ajusta la edad según sea necesario para la definición de 'niño')
# También podemos considerar un rango de edad más específico para educación escolar, por ejemplo, 5 a 17 años.
ninos = pob_ingrefam[pob_ingrefam['edad'] < 18].copy()

# Eliminar filas con valores NaN en las columnas clave para el análisis
ninos_clean = ninos.dropna(subset=['nivel_grado_aprobado', 'ing_familiar', 'est_pobreza']).copy()

print("Descripción del DataFrame de niños después de la fusión y el filtro:")
display(ninos_clean.head())
display(ninos_clean.info())

# Convertir 'est_pobreza' a tipo categórico para un mejor análisis si aún no lo es
ninos_clean['est_pobreza'] = ninos_clean['est_pobreza'].astype('category')

# Analizar la correlación entre `ing_familiar` y `nivel_grado_aprobado`
plt.figure(figsize=(12, 7))
sns.scatterplot(x='ing_familiar', y='nivel_grado_aprobado', data=ninos_clean, alpha=0.6)
plt.title('Correlación entre Ingreso Familiar y Nivel Educativo de Niños')
plt.xlabel('Ingreso Familiar')
plt.ylabel('Nivel y Grado Aprobado')
plt.xscale('log') # Puede ayudar a visualizar la dispersión
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

````

````python
# Calcular el coeficiente de correlación de Pearson
correlation_ing_edu = ninos_clean['ing_familiar'].corr(ninos_clean['nivel_grado_aprobado'])
print(f"\nCoeficiente de correlación de Pearson entre Ingreso Familiar y Nivel Educativo de Niños: {correlation_ing_edu:.2f}")

# Analizar la relación entre `est_pobreza` y `nivel_grado_aprobado`
plt.figure(figsize=(12, 7))
sns.boxplot(x='est_pobreza', y='nivel_grado_aprobado', data=ninos_clean, palette='coolwarm')
plt.title('Nivel Educativo de Niños por Estatus de Pobreza')
plt.xlabel('Estatus de Pobreza (1=Extremo, 2=No Extremo, 3=No Pobre)')
plt.ylabel('Nivel y Grado Aprobado')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Calcular el promedio de nivel educativo por estatus de pobreza
print("\nPromedio del Nivel y Grado Aprobado por Estatus de Pobreza (Niños):")
display(ninos_clean.groupby('est_pobreza')['nivel_grado_aprobado'].mean())

# Conteo de niños por nivel educativo y estatus de pobreza
print("\nDistribución de Nivel Educativo por Estatus de Pobreza (Niños):")
display(pd.crosstab(ninos_clean['est_pobreza'], ninos_clean['nivel_grado_aprobado']))
````

### Pobreza vs. Variables de Educación (`pob`)

Exploraremos la relación entre el estatus de pobreza y variables educativas, como `sabe_leer_escribir` y `nivel_grado_aprobado`.

Para esta parte, necesitaremos unir los dataframes `viv` y `pob` para poder analizar la relación entre características de la población (educación) y la pobreza a nivel de vivienda.

````python
# Unir 'pob' con 'viv' en base a 'upm', 'vivienda' y 'hogar' para obtener el estatus de pobreza
pob_viv = pd.merge(pob, viv[['upm', 'vivienda', 'hogar', 'est_pobreza']], on=['upm', 'vivienda', 'hogar'], how='left')

# Define the poverty labels for the legend
poverty_labels = {1: 'Extremo', 2: 'No Extremo', 3: 'No Pobre'}

# Relación entre 'sabe_leer_escribir' y 'est_pobreza'
plt.figure(figsize=(10, 6))
sns.countplot(x='sabe_leer_escribir', hue='est_pobreza', data=pob_viv.dropna(subset=['sabe_leer_escribir']), palette='plasma')
plt.title('Estatus de Pobreza por Capacidad de Leer y Escribir')
plt.xlabel('Sabe Leer y Escribir (1=Sí, 2-6=No por distintas razones)')
plt.ylabel('Número de Personas')
handles, current_labels = plt.gca().get_legend_handles_labels()
ordered_labels = [poverty_labels[int(label)] for label in current_labels]
plt.legend(handles=handles, labels=ordered_labels, title='Estatus de Pobreza')
plt.show()

# Relación entre 'nivel_grado_aprobado' y 'est_pobreza'
# Para mejor visualización, podemos agrupar 'nivel_grado_aprobado' o tomar un subconjunto.
# Aquí, mostraremos el promedio de 'nivel_grado_aprobado' por estatus de pobreza.
print("\nPromedio de Nivel de Grado Aprobado por Estatus de Pobreza:")
display(pob_viv.groupby('est_pobreza')['nivel_grado_aprobado'].mean())

plt.figure(figsize=(15, 7))
sns.boxplot(x='est_pobreza', y='nivel_grado_aprobado', data=pob_viv.dropna(subset=['nivel_grado_aprobado']), palette='mako', hue='est_pobreza', legend=False)
plt.title('Nivel y Grado Aprobado por Estatus de Pobreza')
plt.xlabel('Estatus de Pobreza (1=Extremo, 2=No Extremo, 3=No Pobre)')
plt.ylabel('Nivel y Grado Aprobado')
plt.show()
````
