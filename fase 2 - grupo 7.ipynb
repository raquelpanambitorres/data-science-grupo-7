# Inferencias Estadísticas: Impacto del Nivel Socioeconómico en la Educación de los Niños

En esta sección evaluaremos estadísticamente si el nivel socioeconómico de los hogares paraguayos (medido a través del **Ingreso Familiar** y el **Estatus de Pobreza**) afecta significativamente al **Nivel Educativo Aprobado** por los niños (menores de 18 años).

## Hipótesis Planteadas:
*   **Hipótesis 1 (Asociación Continua):**
    *   **H0:** No existe correlación entre el ingreso
    (`ing_familiar`) y el nivel educativo aprobado (`nivel_grado_aprobado`) de los niños.
    *   **H1:** Existe correlación entre el ingreso familiar  (`ing_familiar`) y el nivel educativo aprobado (`nivel_grado_aprobado`) de los niños.
*   **Hipótesis 2 (Comparación de Grupos):**
    *   **H0:** El nivel educativo de los niños es igual independientemente del estatus de pobreza del hogar
    *   **H1:** El nivel educativo de los niños difiere significativamente según el estatus de pobreza del hogar.
    (`est_pobreza`).

````python
# Usaremos el conjunto de datos de niños limpio que definimos previamente: `ninos_clean`
# Recordatorio de variables:
# - 'ing_familiar' (Socioeconómico continuo)
# - 'est_pobreza' (Socioeconómico categórico: 1=Extremo, 2=No Extremo, 3=No Pobre)
# - 'nivel_grado_aprobado' (Educativo continuo/ordinal)

print(f"Cantidad de registros para el análisis de niños: {len(ninos_clean)}")
````

## 1. Verificación de Supuestos

Para decidir si usamos pruebas paramétricas (Pearson, ANOVA) o no paramétricas (Spearman, Kruskal-Wallis), debemos verificar los supuestos de **normalidad** y **homocedasticidad** (igualdad de varianzas).

````python
# 1.1 Prueba de Normalidad (Kolmogorov-Smirnov para muestras grandes)
# Evaluamos el nivel educativo y el ingreso familiar logarítmico

ks_edu = stats.kstest(ninos_clean['nivel_grado_aprobado'], 'norm', args=(ninos_clean['nivel_grado_aprobado'].mean(), ninos_clean['nivel_grado_aprobado'].std()))
ks_ing = stats.kstest(np.log1p(ninos_clean['ing_familiar']), 'norm', args=(np.log1p(ninos_clean['ing_familiar']).mean(), np.log1p(ninos_clean['ing_familiar']).std()))

print("--- Prueba de Normalidad (Kolmogorov-Smirnov) ---")
print(f"Nivel Educativo Aprobado: p-valor = {ks_edu.pvalue:.4f} (¿Es normal?: {ks_edu.pvalue > 0.05})")
print(f"Logaritmo del Ingreso Familiar: p-valor = {ks_ing.pvalue:.4f} (¿Es normal?: {ks_ing.pvalue > 0.05})")

# 1.2 Prueba de Homogeneidad de Varianzas (Levene) por Estatus de Pobreza
grupos_edu = [
    ninos_clean[ninos_clean['est_pobreza'] == 1]['nivel_grado_aprobado'],
    ninos_clean[ninos_clean['est_pobreza'] == 2]['nivel_grado_aprobado'],
    ninos_clean[ninos_clean['est_pobreza'] == 3]['nivel_grado_aprobado']
]

stat_levene, p_levene = stats.levene(*grupos_edu)
print("\n--- Prueba de Homocedasticidad (Levene) ---")
print(f"Nivel Educativo entre grupos de pobreza: p-valor = {p_levene:.4f} (¿Varianzas iguales?: {p_levene > 0.05})")
````

Debido a que las variables no siguen una distribución estrictamente normal (p-valor < 0.05) y las varianzas no son homogéneas, procederemos con **pruebas no paramétricas** robustas:
*   **Prueba de Correlación de Spearman** para evaluar la relación entre el Ingreso Familiar y el Nivel Educativo.
*   **Prueba de Kruskal-Wallis** para comparar el Nivel Educativo según el Estatus de Pobreza.

````python
# 2. Ejecución de Pruebas No Paramétricas con Interpretación de Hipótesis Invertidas

# 2.1 Correlación de Spearman
coef_spearman, p_val_spearman = stats.spearmanr(ninos_clean['ing_familiar'], ninos_clean['nivel_grado_aprobado'])
print("--- Resultado de la Correlación de Spearman ---")
print(f"Coeficiente de correlación: {coef_spearman:.4f}")
print(f"p-valor: {p_val_spearman:e}")

# Lógica de decisión: p-valor < 0.05 implica rechazar H0 (Existe correlación) a favor de H1 (No existe correlación)
if p_val_spearman < 0.05:
    print("Resultado: Se rechaza la hipótesis nula (H0). No existe correlación significativa entre el ingreso familiar y el nivel educativo (se apoya H1).")
else:
    print("Resultado: No se puede rechazar la hipótesis nula (H0). Se mantiene la afirmación de H0: Existe una correlación estadísticamente significativa.")

# 2.2 Prueba de Kruskal-Wallis
stat_kw, p_val_kw = stats.kruskal(*grupos_edu)
print("\n--- Resultado de la Prueba de Kruskal-Wallis ---")
print(f"Estadístico H: {stat_kw:.4f}")
print(f"p-valor: {p_val_kw:e}")

# Lógica de decisión: p-valor < 0.05 implica rechazar H0 (El nivel educativo difiere) a favor de H1 (El nivel es igual)
if p_val_kw < 0.05:
    print("Resultado: Se rechaza la hipótesis nula (H0). El nivel educativo es igual independientemente del estatus de pobreza (se apoya H1).")
else:
    print("Resultado: No se puede rechazar la hipótesis nula (H0). Se mantiene la afirmación de H0: El nivel educativo difiere significativamente según el estatus de pobreza del hogar.")
````

### Análisis Post-Hoc (Prueba de Dunn)
Dado que el p-valor de la prueba de Kruskal-Wallis (`0.2156`) es mayor que $0.05$, no se observa evidencia matemática de diferencias en los rangos de nivel educativo aprobado entre los grupos de pobreza. No obstante, por rigor analítico, realizaremos la prueba de Dunn para comprobar si en efecto las comparaciones por parejas confirman la ausencia de diferencias estadísticamente significativas.

## Conclusiones Científicas (Marco de Hipótesis Tradicional)

Tras realizar un análisis estadístico riguroso utilizando pruebas no paramétricas (debido a la falta de normalidad y homocedasticidad en la distribución de los datos), se concluye lo siguiente:

1. **Validación de Supuestos:**
   * Las pruebas de Kolmogorov-Smirnov y Levene confirmaron que las variables educativas y económicas no siguen una distribución normal ni presentan varianzas homogéneas ($p < 0.05$). Esto validó metodológicamente la necesidad de utilizar pruebas no paramétricas robustas en lugar de pruebas paramétricas clásicas.

2. **Relación entre Ingreso Familiar y Nivel Educativo (Hipótesis 1):**
   * **Hipótesis:**
     * $H_0$: No existe correlación entre el ingreso familiar y el nivel educativo aprobado por los niños.
     * $H_1$: Existe correlación entre el ingreso familiar y el nivel educativo aprobado por los niños.
   * **Resultado:** La prueba de correlación de Spearman arrojó un coeficiente de correlación extremadamente bajo de `0.0057` con un p-valor de `0.5136`.
   * **Decisión:** Al ser el p-valor muy superior al nivel de significancia estándar ($\alpha = 0.05$), **no se puede rechazar la hipótesis nula ($H_0$)**.
   * **Conclusión:** No existe evidencia estadística que demuestre una correlación significativa entre el ingreso mensual del hogar y el nivel educativo aprobado por los niños en la muestra analizada.

3. **Diferencias según Estatus de Pobreza (Hipótesis 2):**
   * **Hipótesis:**
     * $H_0$: El nivel educativo de los niños es igual independientemente del estatus de pobreza del hogar.
     * $H_1$: El nivel educativo de los niños difiere significativamente según el estatus de pobreza del hogar.
   * **Resultado:** La prueba de Kruskal-Wallis arrojó un estadístico $H = 3.0684$ con un p-valor de `0.2156`.
   * **Decisión:** Como el p-valor es mayor a $0.05$, **no se puede rechazar la hipótesis nula ($H_0$)**.
   * **Conclusión:** No se observan diferencias estadísticamente significativas en el nivel educativo aprobado por los niños según el estatus de pobreza de su hogar (pobreza extrema, pobreza no extrema y no pobre).

4. **Confirmación Post-Hoc:**
   * La prueba de Dunn con ajuste de Holm confirmó este escenario, mostrando p-valores de comparación por parejas consistentes por encima de $0.05$ (todos en el orden de `0.4009`), corroborando que las diferencias entre los subgrupos son puramente debidas al azar muestral.
